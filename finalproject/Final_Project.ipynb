```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
import pandas as pd
from ipywidgets import interact
import math as math
from astropy import units as u
from tkinter import PhotoImage
from tkinter import *
from PIL import ImageTk, Image
import os
```

```python
from tkinter import * 

#starting the tkinter tab
root = Tk()  

#setting a bound (frame) for the bottom of the tkinter tab so that the sliders remain on screen
bottom_frame = Frame(root)
bottom_frame.pack(side="bottom", fill="both", expand=True)
```

```python
import PIL.Image

#Lsun is the luminosity of our sun, which will be used as a baseline to calculate other planet's habitability
Lsun = 5778

#creating another bound (frame) so that both the planet image and the star image can coexist without overlapping on each other
planet_panel = Label(bottom_frame)
planet_panel.pack(side="right", fill="both", expand=True, padx=10)


def display_habitability(Luminosity, Distance):
    """
    This function uses the Predicted Observables for Exoplanets Tool from NASA exoplanet archive to calculate the ranges of habitable exoplanets based on our solar's system's ranges of habitable planets
    
    -------Parameters-----------------------
    Luminosity=Lval: float
    Lsun: Luminosity of our sun: float
    Distance in AU: float

    ----------Returns----------------------------
    displays photos of each on 5 categories of planets relative to habitability: Dry, Inner_Habitable, Habitable, Outer_Habitable, Snowball
    """

    #Is used to allow the sliders to take values from the function
    Lval= Luminosity.get()
    Dval = Distance.get()

    #the functions to calculate Habitable Zone from NASA exoplanet archive
    RHZinner = 0.75 * math.sqrt(Lval/Lsun) 
    RHZouter = 1.77 * math.sqrt(Lval/Lsun) 
    RHZcenter = 1.0 * math.sqrt(Lval/Lsun) 
    RHZcenter_inner = RHZcenter + RHZinner
    RHZcenter_outer = RHZcenter + RHZouter


    #Detects if its within the broad habitable range (including inner and outer) or not
    
    
    #not within the habitable range, so its seeing if its a dry planet or if its a snowball
    if Dval < RHZinner:
        #display Dry_planet
        img_file = "Dry_Planet.png"
    elif Dval > RHZouter:
        #display Snowball_planet
        img_file = "Snowball_Planet.png"

        
    else:
        #if Dval >=RHZinner and Dval <=RHZouter:
        #displays based on which set of ranges of habitability it falls within
        if Dval >=RHZinner and Dval <=RHZcenter_inner/2:
            img_file = "Inner_Habitable_Planet.png"
                #display Inner_Habitable.png
        elif Dval >= RHZcenter_outer/2 and Dval<=RHZouter:
            img_file = "Outer_Habitable_Planet.png"
                #display Outer_Habitable.png
        else: 
                #display Habitable_Planet.png
            img_file = "Habitable_Planet.png"
            
    #helps configure the images so that they're able to be imported into the tkinter file
    raw_img = PIL.Image.open(img_file)
    img = ImageTk.PhotoImage(raw_img)
    
    planet_panel.config(image=img)
    planet_panel.image = img


```

```python
#creating another bound (frame) so that both the planet image and the star image can coexist without overlapping on each other
star_panel = Label(bottom_frame)
star_panel.pack(side="left", fill="both", expand=True, padx=10)

def display_star(Luminosity):

    """
    Determines what spectral class the star is based on what luminosity it is.

    -----------Parameters--------------
    luminosity: float
    ----------Returns----------------------------
    displays photos of each on 7 categories of planets relative to luminosity and heat

    """
    #lets the luminosity value be taken from the slider
    val= Luminosity.get()

    #displays image based on the range of luminosity
    if val>=2000 and val<3500:
        #display M Star
        img_file = "M_star.png"
    elif val>=3500 and val<5000:
        #display K Star
        img_file = "K_star.png"
    elif val>=5000 and val<6000:
        #display G Star
        img_file = "G_star.png"
    elif val>=6000 and val<7500:
        #display F Star
        img_file = "F_star.png"
    elif val>=7500 and val<10000:
        #display A Star
        img_file = "A_star.png"
    elif val>=10000 and val<30000:
        #display B Star
        img_file = "B_star.png"
    else:
        #O_star = true
        #display O Star
        img_file = "O_star.png"
    #helps configure the images so that they're able to be imported into the tkinter file
    raw_img = PIL.Image.open(img_file)
    img = ImageTk.PhotoImage(raw_img)
    
    star_panel.config(image=img)
    star_panel.image = img
```

```python


#creating the size of the window
root.geometry("1200x900") 

v1 = DoubleVar()
v2 = DoubleVar()
def show1():  
    
    sel = "Horizontal Scale Value = " + str(v1.get())
    l1.config(text = sel, font =("Courier", 14))  



#creating a frame for the sliders so that they don't leave screen when the other images are introduced
top_frame = Frame(root)
top_frame.pack(side="top", fill="x")

#creating the slider for the luminosity input
s1 = Scale( top_frame, variable = v1, 
           from_ = 2000, to = 60000, 
           orient = HORIZONTAL, resolution = 0.01, 
            #linking the slider to the function display_star()
            command=lambda x: display_star(v1)
            )

#creating the label and the position of the slider on the screen
l3 = Label(top_frame, text = "Luminosity in Kelvin")
l1 = Label(top_frame)
s1.pack(anchor = CENTER) 
l3.pack()
l1.pack() 


#creating the slider for the distance input
s2 = Scale( top_frame, variable = v2, 
           from_ = 0.01, to = 20, resolution = 0.1,
            #linking the slider to the function display_habitability()
           orient = HORIZONTAL,command=lambda x: display_habitability(v1,v2))
            
#creating the label and the position of the slider on the screen
l4 = Label(top_frame, text = "Distance in AU")
l2 = Label(top_frame)
s2.pack(padx = 100)
l4.pack()
l2.pack() 




root.mainloop()
```
