```python
#2.2.
#1
mass_unit = 1e24 * u.kg
dist_unit = 1e6 * u.km
def attach_units(df, old_col, unit, new_col=None):
    new_values = []
    for val in df[old_col]:
        new_values.append(val * unit)
    df[old_col] = new_values
    if new_col:
        df.rename(columns={old_col: new_col}, inplace=True)
```

```python
hours_unit = 1 * u.h
hour_column = attach_units(new_df, "Length of Day (hours)", hours_unit, "Length of Day in Hours")
```

```python
#2.3
Perihelion = new_df['Perihelion (10^6 km)']
Aphelion = new_df['Aphelion (10^6 km)']
semi_major = (Perihelion + Aphelion) / 2

new_df.insert(11, "Semi-Major Axis (10^6 km)", semi_major)
print(new_df["Semi-Major Axis (10^6 km)"])
```
