```python
#4.1 -------------------------
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#4.2---------------
df = pd.read_csv("exoplanet_data.csv", comment = "#")
```

```python
#4.2.4
shape = df.shape
print("Number of Columns: ", shape[1])    #4
print("Number of Rows: ", shape[0])
print(df.columns)
#Titles that I don't understand are: sy_disterr2, pl_radj, st_logg, pl_insol, ttv_flag
```

```python
#4.3-------------------------

#3. Planet Name-pl_name, Planet name most commonly used in the literature, most common name for the planet
#Host Name-hostname, Stellar name most commonly used in the literature, most common name for the star that the planet is in
#Discovery Method-discoverymethod, Method by which the planet was first identified, how the planet was first identified
#Spectral Type- st_spectype, Classification of the star based on their spectral characteristics following the Morgan-Keenan system,a way to classify stars based on their magnitude, temperature and color.
#Equilibrium Temperature [K]- pl_eqt, The equilibrium temperature of the planet as modeled by a black body heated only by its host star, or for directly imaged planets, the effective temperature of the planet required to match the measured luminosity if the planet were a black body, the temperature that a planet needs to have to match lumnosity of a black body based on the amount of heat given by its star.
#Planet Mass or Mass*sin(i) [Jupiter Mass]- pl_bmassj, Best planet mass estimate available, in order of preference: Mass, M*sin(i)/sin(i), or M*sin(i), depending on availability, and measured in Jupiter masses, the mass of a planet based on its relation to the weight of jupiter.
```

```python
#4
data = df[['pl_name', 'hostname', 'discoverymethod', 'st_spectype', 'pl_eqt', 'pl_bmassj']]
data
```

```python
#5
new_data = data.dropna()
new_data
```

```python
#6
new_shape = new_data.shape
print("Number of Columns: ", new_shape[1])    #4
print("Number of Rows: ", new_shape[0])
print(new_data.columns)
```

```python
#there's radial velocity and transit. There are 5 transits, and 2 radial velocities. Radial velocity causes a small wobble on a star due to gravitry that lets scientists know that there's a planet there. The transit method causes a small dip in the level of light when a planet crosses past its star dfrom our perspective that allows us to see it.
```

```python
#2
f, ax1 = plt.subplots(1, figsize=(8, 6), sharex=True)
x = new_data.loc[:, 'pl_bmassj']
y = new_data.loc[:, 'pl_eqt']
ax1.set_xlabel("Planet Mass in Jupiters")
ax1.set_ylabel("Planet Temperature Equibrium")
plt.scatter(x, y,)
```

```python
#3


f, ax1 = plt.subplots(1, figsize=(8, 6), sharex=True)
x = np.log(new_data.loc[:, 'pl_bmassj'])
y = np.log(new_data.loc[:, 'pl_eqt'])
ax1.set_xlabel("Planet Mass in Jupiters")
ax1.set_ylabel("Planet Temperature Equibrium")
t = ['red']
plt.scatter(x, y, c=t)
```
