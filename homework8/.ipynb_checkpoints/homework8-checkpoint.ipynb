```python
#6
# The sepal_length is the length in inches of the green part covering the petals of an iris to protect it.
# Likewise, the sepal_width is the width in inches of the green part covering the petals of an iris to protect it.
# The petal length is how long in inches outwards the petal extends from the bud 
# The petal width is how long in inches widthwise the petal is. 
```

```python
#7
sepal_length = new_df.loc[:, 'sepal_length']
petal_length = new_df.loc[:, 'petal_length']
sepal_width = new_df.loc[:, 'sepal_width']
petal_width = new_df.loc[:, 'petal_width']
```

```python
fig, (ax1, ax2) = plt.subplots(2, 2,)
sns.histplot(petal_width) #bottom left
sns.boxplot(sepal_length) #bottom right
plt.scatter(sepal_length, petal_length) #top right length to length
plt.scatter(sepal_width, petal_width) #top left wisth to width
plt.show()
```

```python
#9.
#I decided on these because subplots because i wanted to compare the length of the petal to the length of the sepal, and the width of the petal to the width of the sepal.
#I could see that the widths of the petals were much longer and spread out to that of the sepals, and likewise for the lengths.
```
